# Step 1 — Test: loader.py

验证 loader.py 能正确读取原始数据。

```python
import sys
sys.path.insert(0, '..')   # 让 Python 能找到 src/ 文件夹

import yaml
from src.data.loader import load_raw_ohlcv
```

```python
# 读取 config
with open('../configs/config.yaml', 'r') as f:
    cfg = yaml.safe_load(f)

print('Config loaded. file_path:', cfg['data_source']['file_path'])
```

```python
# 读取数据
df = load_raw_ohlcv(cfg)
df.head()
```

```python
# 基本检查
print('Shape:', df.shape)
print('Dtypes:\n', df.dtypes)
print('Index timezone:', df.index.tz)
print('Any NaN?', df.isna().any().any())
print('Duplicated timestamps?', df.index.duplicated().any())
```

```python
# 画一小段价格，确认数据正常
import matplotlib.pyplot as plt

sample = df['Close'].iloc[:500]
plt.figure(figsize=(14, 4))
plt.plot(sample.index, sample.values)
plt.title('NQ Close Price — first 500 bars (1min)')
plt.xlabel('Time')
plt.ylabel('Price')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
